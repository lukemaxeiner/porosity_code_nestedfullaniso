```python
import tifffile as tiff

input_file = "S4_1.0487_2201_NI8.tif"
output_file = "S4_1.0487_2201_NI8crop.tif"

crop = 1200

with tiff.TiffFile(input_file) as tif:
    pages = tif.pages
    z = len(pages)
    y, x = pages[0].shape

    z0 = (z - crop) // 2
    y0 = (y - crop) // 2
    x0 = (x - crop) // 2

    z1 = z0 + crop
    y1 = y0 + crop
    x1 = x0 + crop

    if z < crop or y < crop or x < crop:
        raise ValueError(f"Image too small: found {z}x{y}x{x}, need 1200x1200x1200")

    with tiff.TiffWriter(output_file, bigtiff=True) as out:
        for i in range(z0, z1):
            slice_2d = pages[i].asarray()
            cropped = slice_2d[y0:y1, x0:x1]
            out.write(cropped, photometric="minisblack")

print("Saved:", output_file)
```
